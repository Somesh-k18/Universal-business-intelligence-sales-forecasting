## 📈 Exploratory Analysis..

Purpose: Analyze trends and generate business insights.

1.Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

2.Load Clean Dataset

```python
df = pd.read_csv("cleaned_sales_data.csv")
```

3.Total Revenue

```python
df["Revenue"].sum()
```

4.Monthly Sales Trend

```python
monthly_sales = df.groupby("Month_Name")["Revenue"].sum()

monthly_sales.plot(
    kind="line",
    marker="o",
    figsize=(10,5)
)

plt.title("Monthly Revenue Trend")
plt.show()
```

5.Category-wise Sales

```python
category_sales = df.groupby("Category")["Revenue"].sum()

category_sales.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Category-wise Revenue")
plt.show()
```

6.Region-wise Revenue

```python
region_sales = df.groupby("Region")["Revenue"].sum()

plt.figure(figsize=(8,5))
sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

plt.title("Revenue by Region")
plt.show()
```

7.Top Products

```python
top_products = df.groupby("Product")["Revenue"].sum()

top_products = top_products.sort_values(
    ascending=False
).head(10)

top_products.plot(
    kind="bar"
)

plt.title("Top 10 Products")
plt.show()
```

8.Revenue Distribution

```python
sns.histplot(df["Revenue"], bins=20)

plt.title("Revenue Distribution")
plt.show()
```

9.Correlation Heatmap

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(8,5))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="Blues"
)

plt.show()
```

### 10.Business Insights


• Accessories generated highest revenue.

• South region performed best.

• Sales increased significantly in Q4.

• Top 10 products contribute major revenue.
