### 📒 Data Cleaning..

Purpose: Convert raw sales data into a clean dataset for analysis and forecasting.

1. Import Libraries

```python
import pandas as pd
import numpy as np
```

2. Load Dataset

```python
df = pd.read_csv("../Sample_Datasets/sales_data.csv")
df.head()
```

3. Check Dataset Information

```python
df.info() 
df.shape
df.describe()
```

4. Check Missing Values

```python
df.isnull().sum()
```

5. Remove Duplicate Records

```python
df.duplicated().sum()

df.drop_duplicates(inplace=True)
```

6. Extract Date Features

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

print(df["Date"].dtype)
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()
df["Quarter"] = df["Date"].dt.quarter
```

7. Save Clean Dataset

```python
df.to_csv("cleaned_sales_data.csv", index=False)
```
